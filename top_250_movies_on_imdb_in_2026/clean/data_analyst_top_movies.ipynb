## jumlah genre terbanyak

menghitung jumlah genre film terbanyak

```python
def parse_genres(value):
    if pd.isna(value):
        return []
    
    # Jika sudah berupa list
    if isinstance(value, list):
        return value
    
    # Jika berupa string
    if isinstance(value, str):
        try:
            result = ast.literal_eval(value)
            
            if isinstance(result, list):
                return result
            
            return [str(result)]
        
        except (ValueError, SyntaxError):
            # Jika format bukan Python list,
            # pisahkan berdasarkan koma
            return [x.strip() for x in value.split(',') if x.strip()]
    
    return []
```

```python
df['genres'] = df['genres'].apply(parse_genres)
df[['primaryTitle', 'genres']].head(10)
```

```python
genre_df = df[
    ['primaryTitle', 'genres', 'averageRating']
].explode('genres')
genre_counts = genre_df['genres'].value_counts()

genre_counts.head(10)
```

## rata rata rating berdasarkan genre

```python
genre_rating = (
    genre_df
    .groupby('genres')['averageRating']
    .agg(['count', 'mean'])
    .sort_values('mean', ascending=False)
)
genre_rating.columns = ['Movie Count', 'Average Rating']

genre_rating_filtered = genre_rating[
    genre_rating['Movie Count'] >= 5
].sort_values(
    'Average Rating',
    ascending=False
)

genre_rating_filtered
```

## tahun dengan jumlah film terbanyak

### dihitung pertahun

```python
movies_by_year = (
    df['Year']
    .value_counts()
    .sort_index()
)
movies_by_year.sort_values(
    ascending=False
).head(10)
```

### dihitung per decade

```python
df['decade'] = (
    (df['Year'] // 10) * 10
)
movies_by_decade = (
    df['decade']
    .value_counts()
    .sort_index()
)

movies_by_decade
```

## ringkasan

```python
decade_summary = (
    df.groupby('decade')
    .agg(
        Movie_Count=('primaryTitle', 'count'),
        Average_Rating=('averageRating', 'mean'),
        Average_Runtime=('runtimeMinutes', 'mean'),
        Average_Votes=('numVotes', 'mean')
    )
    .round(2)
)

decade_summary
```
